### Passo a passo para tratamento dos dados baixados do Kaggle

```python
# Passo 1 - Importação de bibliotecas utilizadas.
import pandas as pd
import random as ra 
import os 
```

```python
# Passo 2 -  Carregando o arquivo .csv
df = pd.read_csv('dados_analise_fraude.csv')
```

```python
# Passo 3 -  Visualiza do Data frame para garantir que importou.
df.head()
```

```python
# Passo 4 - Entendendo os dados.
# Para esse passo execute cada linha de uma  vez

#df.shape # Mostra a quantidade de linhas e colunas de uma base de dados 
print(df.columns) # Mosta os nomes das colunas 
#df.info() # Mosta os tipos dessas colunas.
#df.describe #Resumo estatístico dos dados.

```

```python
# Passo 5 - Alterando os nomes das colunas para facilitar o entendimento dos dados.
#  ** Como temos aqui um projeto pessoal isso não é um problema no mundo real qualquer alteração de nome  e ou tipo deve ser previamente acordada e documentada com todos os envolvidos

df = df.rename(columns={
    'step': 'TEMPO_DECORRIDO',
    'type': 'TIPO_TRANSACAO',
    'amount': 'VALOR',
    'nameOrig': 'ID_CLIENTE',
    'oldbalanceOrg': 'SALDO_CLIENTE_ANTERIOR',
    'newbalanceOrig': 'SALDO_POSTERIOR',
    'nameDest': 'ID_ESTABELECIMENTO',
    'oldbalanceDest': 'SALDO_ESTABELECIMENTO_ANTERIOR',
    'newbalanceDest': 'SALDO_ESTABELECIMENTO_POSTERIOR',
    'isFraud': 'TRANSACAO_FRAUDULENTA',
    'isFlaggedFraud': 'MARCADA_POR_REGRA'
    })
```

```python
# Passo 6 -  Criando a coluna UF_ESTABELECIMENTO para deixar a base mais interessante.
# **  Lembrando que se você rodar esse código mais de uma vez o sorteio de  UFs acontece novamente alterando a UF de cada um dos estabelecimentos.



# Lista contendo todos os estados Brasileiros.
ufs = [
    'AC', 'AL', 'AP', 'AM', 'BA', 'CE', 'DF',
    'ES', 'GO', 'MA', 'MT', 'MS', 'MG', 'PA',
    'PB', 'PR', 'PE', 'PI', 'RJ', 'RN', 'RS',
    'RO', 'RR', 'SC', 'SP', 'SE', 'TO'
]

# Mapeia os ids de estabelecimento como únicos para não ter um estabelecimento com mais de um estado 

estabelecimentos = df['ID_ESTABELECIMENTO'].unique()
mapa_ufs = {
    estabelecimento: ra.choice(ufs)
    for estabelecimento in estabelecimentos
}

# Adiciona a nova coluna ao DataFrame
df['UF_ESTABELECIMENTO'] = df['ID_ESTABELECIMENTO'].map(mapa_ufs)
```

```python
# Passo 7 - Trata o conteúdo da coluna TIPO_TRANSAÇÃO
# Como não temos duplicados, valores inválidos ou  Nulos faremos somente a tradução.

df['TIPO_TRANSACAO'] = df['TIPO_TRANSACAO'].replace({
    'PAYMENT': 'PAGAMENTO',
    'TRANSFER': 'TRANSFERENCIA',
    'CASH_OUT': 'SAQUE',
    'CASH_IN': 'DEPOSITO',
    'DEBIT': 'DEBITO'
})
```

```python
# Passo 8 - Exportando o Data Frame para CSV novamente para importar no Postgres.

# Exporta o data frame para csv.
df.to_csv('dados_analise_fraude_v2.csv', index=False, encoding='utf-8')
print("Arquivo exportado com sucesso!")

# Mostra onde foi salvo.
print(os.path.abspath('dados_analise_fraude_v2.csv'))
```
